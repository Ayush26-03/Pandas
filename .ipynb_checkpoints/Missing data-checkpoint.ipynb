```python
import numpy as np
import pandas as pd
```

```python
###  MISSING DATA
```

```python
###  1. Finding Missing data
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

```python
#### 2. Removing Null
```

```python
df.dropna()
```

```python
df.dropna(thresh=3)
```

```python
### 3. Filling missing Data
```

```python
df.fillna(0)
```

```python
Dict={"A":100,"B":200,"C":300,"D":400}
df.fillna(value=Dict)
```
